```python
import json
import os
import random
import re
import time
import unicodedata
from concurrent.futures import ThreadPoolExecutor, as_completed
from difflib import SequenceMatcher

import pandas as pd
from google_play_scraper import search, app as app_details
```

```python
INPUT_CSV = "./data/google_play_store_cleaned.csv"
OUTPUT_CSV = "./data/google_play_store_cleaned_enriched.csv"
CACHE_PATH = "./data/app_icons.v3.cache.jsonl"       
OLD_CACHE_PATHS = ["./data/app_icons.v2.cache.jsonl", "./data/app_icons.cache.jsonl"]   
REUSE_OLD_CACHE = False  
DETAILS_CACHE_PATH = "./data/app_details.cache.jsonl"  
WORKERS = 4
MIN_SCORE = 0.8
LIMIT = None          # None = process EVERY app; set a number (e.g. 50) only for a quick test
CHUNK_SIZE = 250      # apps per chunk: each chunk is fetched, cached to disk and checkpointed before the next starts
CHUNK_PAUSE = 3       # seconds to rest between chunks (avoids rate limits)
FETCH_DETAILS = True  # also pull full store data (developer, description, histogram, ...) per matched app
DETAIL_STATUSES = ("ok",)  # add "low_confidence" to also fetch details for uncertain matches
USE_CATEGORY_HINT = True   # also try "<app name> <category>" as a query (helps generic / short names)
LANG_COUNTRY_TRIES = [("en", "us"), ("fr", "fr"), ("de", "de")]  # (lang, country) markets to search; names like "...expliqué" / "Schlaumeier" are non-English
EARLY_STOP_SCORE = 0.95    # stop trying more queries once a match this close is found
# how the match is verified (candidate score = title similarity + bonuses/penalties below)
CATEGORY_BONUS = 0.05      # candidate's Play category agrees with the dataset category
PRICE_TYPE_BONUS = 0.03    # candidate is free/paid like the dataset says
RATING_BONUS = 0.03        # candidate's rating within 0.3 of the dataset rating
NUMBER_MISMATCH_PENALTY = 0.25  # model numbers differ, e.g. "CZ-70" vs "CZ-50" (stops near-identical names matching the wrong app)
AMBIGUITY_MARGIN = 0.03    # flag as ambiguous when the runner-up is within this score of the winner
```

```python
import re
import unicodedata

def norm(s: str) -> str:
    """lowercase, strip accents/punctuation, collapse spaces"""
    s = unicodedata.normalize("NFKD", str(s))
    s = "".join(ch for ch in s if not unicodedata.combining(ch))
    return re.sub(r"\s+", " ", "".join(ch.lower() if ch.isalnum() else " " for ch in s)).strip()

def score(a: str, b: str) -> float:
    """title similarity 0-1: best of character similarity and word-set overlap (robust to word order / extra words)"""
    na, nb_ = norm(a), norm(b)
    if not na or not nb_:
        return 0.0
    seq = SequenceMatcher(None, na, nb_).ratio()
    ta, tb = set(na.split()), set(nb_.split())
    overlap = len(ta & tb) / max(len(ta), len(tb))
    return max(seq, overlap)

def number_tokens(s: str) -> set:
    """digit groups in a name (model numbers, versions): 'CZ-70 (CZ-50)' -> {'70', '50'}"""
    return set(re.findall(r"\d+", norm(s)))

def cat_key(c) -> str:
    """dataset category 'Books & Reference' -> 'BOOKS_AND_REFERENCE' (Play genreId style)"""
    return re.sub(r"[^A-Z]+", "_", str(c).upper().replace("&", "AND")).strip("_")

def score_candidate(app_name: str, hit: dict, ctx: dict) -> tuple:
    """Return (final_score, title_score, signals). Signals explain why a candidate won."""
    t = score(app_name, hit.get("title", ""))
    final, signals = t, []
    # 1) model numbers must agree when both names have them
    na, nh = number_tokens(app_name), number_tokens(hit.get("title", ""))
    if na and nh and na != nh:
        final -= NUMBER_MISMATCH_PENALTY; signals.append("numbers_differ")
    elif na and na == nh:
        signals.append("numbers_match")
    # 2) category agreement (hit genreId e.g. GAME_ACTION agrees with dataset category 'Game')
    cat, gid = ctx.get("category"), hit.get("genreId")
    if cat and gid and (str(gid).startswith(cat_key(cat)) or cat_key(cat).startswith(str(gid))):
        final += CATEGORY_BONUS; signals.append("category")
    # 3) free / paid agreement
    typ = ctx.get("type")
    if typ and hit.get("free") is not None and (typ == "Free") == bool(hit["free"]):
        final += PRICE_TYPE_BONUS; signals.append("price_type")
    # 4) rating proximity
    r, hs = ctx.get("rating"), hit.get("score")
    if r is not None and hs is not None and r == r and abs(float(r) - float(hs)) <= 0.3:
        final += RATING_BONUS; signals.append("rating")
    return final, t, signals

def query_variants(app_name: str, category=None) -> list:
    """Different ways to phrase the same app name, most specific first."""
    name = str(app_name).strip()
    no_paren = re.sub(r"\s*[\(\[].*?[\)\]]", "", name).strip()       # 'Cossack Dictionary (DN)' -> 'Cossack Dictionary'
    in_paren = " ".join(re.findall(r"[\(\[](.*?)[\)\]]", name))
    no_accents = "".join(c for c in unicodedata.normalize("NFKD", name) if not unicodedata.combining(c))
    cands = [name, f'"{name}"', no_paren, no_accents, norm(name), f"{no_paren} {in_paren}".strip()]
    if USE_CATEGORY_HINT and category and isinstance(category, str):
        cands.append(f"{no_paren or name} {category.replace('_', ' ').lower()}")
    seen, out = set(), []
    for q in cands:
        if q and q not in seen:
            seen.add(q)
            out.append(q)
    return out

def lookup(app_name: str, min_score: float, ctx: dict = None, retries: int = 3) -> dict:
    """Step 1: search the Play Store by app name (several phrasings x several markets), then verify every
    candidate against what we know about the app (name, model numbers, category, free/paid, rating)."""
    ctx = ctx or {}
    row = {
        "app": app_name, "matched_title": None, "app_id": None, "icon_url": None,
        "match_score": None, "title_score": None, "match_signals": None, "matched_query": None,
        "match_method": None, "runner_up_title": None, "match_status": "not_found",
    }
    cands, had_error = {}, None          # app_id -> (final, title_score, signals, hit, query)
    for q in query_variants(app_name, ctx.get("category")):
        for lang, country in LANG_COUNTRY_TRIES:
            hits = None
            for attempt in range(retries):
                try:
                    hits = search(q, n_hits=10, lang=lang, country=country)
                    break
                except Exception as e:
                    had_error = type(e).__name__
                    time.sleep(2**attempt + random.random())
            for h in hits or []:
                final, t, sig = score_candidate(app_name, h, ctx)
                if h["appId"] not in cands or final > cands[h["appId"]][0]:
                    cands[h["appId"]] = (final, t, sig, h, q)
        if cands and max(v[0] for v in cands.values()) >= EARLY_STOP_SCORE:
            break
    if cands:
        ranked = sorted(cands.values(), key=lambda v: v[0], reverse=True)
        final, t, sig, hit, q = ranked[0]
        exact = norm(hit["title"]) == norm(app_name)
        ambiguous = len(ranked) > 1 and ranked[1][1] >= min_score and (final - ranked[1][0]) < AMBIGUITY_MARGIN and not exact
        if "numbers_differ" in sig:
            status = "low_confidence"
        elif t >= min_score:
            status = "ambiguous" if ambiguous else "ok"
        else:
            status = "low_confidence"
        row.update(
            matched_title=hit["title"], app_id=hit["appId"], icon_url=hit["icon"],   # always recorded
            match_score=round(min(final, 1.0), 3), title_score=round(t, 3), match_signals=",".join(sig),
            matched_query=q, match_method="exact_title" if exact else "fuzzy_title",
            runner_up_title=ranked[1][3]["title"] if len(ranked) > 1 else None, match_status=status,
        )
    elif had_error:
        row["match_status"] = f"error: {had_error}"
    return row


# fields kept from google_play_scraper.app(); renamed with a store_ prefix so they never clash with dataset columns
DETAIL_FIELDS = {
    "developer": "store_developer",
    "developerId": "store_developer_id",
    "developerWebsite": "store_developer_website",
    "summary": "store_summary",
    "description": "store_description",
    "genre": "store_genre",
    "genreId": "store_genre_id",
    "contentRating": "store_content_rating",
    "score": "store_rating",
    "ratings": "store_rating_count",
    "reviews": "store_review_count",
    "installs": "store_installs_label",
    "realInstalls": "store_real_installs",
    "free": "store_is_free",
    "price": "store_price",
    "currency": "store_currency",
    "offersIAP": "store_offers_iap",
    "adSupported": "store_ad_supported",
    "containsAds": "store_contains_ads",
    "released": "store_released",
    "lastUpdatedOn": "store_last_updated",
    "version": "store_version",
    "androidVersion": "store_min_android",
    "headerImage": "store_header_image",
    "url": "store_url",
}

def fetch_details(app_id: str, retries: int = 3) -> dict:
    """Step 2: full store listing for one app_id (one row, flat columns)."""
    row = {"app_id": app_id, "details_status": "not_found"}
    for attempt in range(retries):
        try:
            d = app_details(app_id, lang="en", country="us")
            for src, dst in DETAIL_FIELDS.items():
                row[dst] = d.get(src)
            hist = d.get("histogram") or [None] * 5   # star counts: 1..5
            for i, v in enumerate(hist, 1):
                row[f"store_stars_{i}"] = v
            row["details_status"] = "ok"
            return row
        except Exception as e:
            msg = type(e).__name__
            if msg == "NotFoundError":          # app removed / region locked: no point retrying
                return row
            row["details_status"] = f"error: {msg}"
            time.sleep(2**attempt + random.random())
    return row
```

```python
# 1. Load existing cleaned dataset
clean_data = pd.read_csv(INPUT_CSV)
apps = clean_data["app"].dropna().drop_duplicates().tolist()

if LIMIT:
    apps = apps[:LIMIT]
# what we know about each app, used to verify candidates (category, free/paid, rating)
_ctx_cols = [c for c in ["category", "type", "rating"] if c in clean_data.columns]
context_by_app = clean_data.drop_duplicates("app").set_index("app")[_ctx_cols].to_dict("index")
```

```python
# 2. Load caches to resume interrupted runs
done = {}

def _load(path, accept):
    if not os.path.exists(path):
        return
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            r = json.loads(line)
            status_val = r.get("match_status") or r.get("status")
            if status_val in accept:
                r["match_status"] = status_val
                done[r["app"]] = r

if REUSE_OLD_CACHE:
    for p in OLD_CACHE_PATHS:
        _load(p, accept=("ok",))
# current (v3) cache: everything it finished (errors are retried)
_load(CACHE_PATH, accept=("ok", "ambiguous", "low_confidence", "not_found"))

todo = [a for a in apps if a not in done]
print(f"Total unique apps: {len(apps)} | Cached: {len(done)} | To fetch: {len(todo)}")
```

```python
# 3. Fetch missing apps in chunks (every result is cached immediately; a checkpoint CSV is saved after each chunk)
def chunked(seq, size):
    for i in range(0, len(seq), size):
        yield seq[i:i + size]


CHECKPOINT_CSV = OUTPUT_CSV.replace(".csv", ".checkpoint.csv")
LOG_EVERY_N_APPS = 10

t0 = time.time()
n_done = 0
status_counts = {"ok": 0, "low_confidence": 0, "not_found": 0, "error": 0}

if todo:
    n_chunks = -(-len(todo) // CHUNK_SIZE)
    print(
        f"Starting fetch: {len(todo)} apps across {n_chunks} chunks "
        f"(chunk_size={CHUNK_SIZE}, workers={WORKERS})",
        flush=True,
    )
    print("-" * 75, flush=True)

    with open(CACHE_PATH, "a", encoding="utf-8") as cache:
        for ci, chunk in enumerate(chunked(todo, CHUNK_SIZE), 1):
            chunk_t0 = time.time()
            print(f"[Chunk {ci}/{n_chunks}] Submitting {len(chunk)} apps...", flush=True)

            with ThreadPoolExecutor(WORKERS) as pool:
                futures = [
                    pool.submit(lookup, a, MIN_SCORE, context_by_app.get(a))
                    for a in chunk
                ]
                for idx_in_chunk, fut in enumerate(as_completed(futures), 1):
                    r = fut.result()
                    done[r["app"]] = r
                    cache.write(json.dumps(r) + "\n")
                    cache.flush()

                    # Track match status counts
                    st = r.get("match_status") or r.get("status") or "error"
                    if st in status_counts:
                        status_counts[st] += 1
                    else:
                        status_counts["error"] += 1
                        print(f"  [!] Error on '{r['app']}': {st}", flush=True)

                    # Intra-chunk progress update
                    if idx_in_chunk % LOG_EVERY_N_APPS == 0 or idx_in_chunk == len(chunk):
                        current_total = n_done + idx_in_chunk
                        pct = (current_total / len(todo)) * 100
                        print(
                            f"  -> {current_total}/{len(todo)} ({pct:5.1f}%) | "
                            f"ok={status_counts['ok']} low={status_counts['low_confidence']} "
                            f"miss={status_counts['not_found']} err={status_counts['error']} | "
                            f"last: '{r['app'][:28]}' ({st})",
                            flush=True,
                        )

                    time.sleep(0.2)

            n_done += len(chunk)
            pd.DataFrame(list(done.values())).to_csv(CHECKPOINT_CSV, index=False)

            elapsed = time.time() - t0
            chunk_elapsed = time.time() - chunk_t0
            rate = n_done / elapsed if elapsed > 0 else 0
            eta = (len(todo) - n_done) / rate if rate > 0 else 0

            print(
                f"[Chunk {ci}/{n_chunks} SAVED] {n_done}/{len(todo)} total | "
                f"chunk took {chunk_elapsed:.1f}s ({rate:.2f} apps/s) | "
                f"elapsed {elapsed/60:.1f}m | ETA ~{eta/60:.1f}m",
                flush=True,
            )

            if n_done < len(todo):
                print(f"  Pausing {CHUNK_PAUSE}s before next chunk...", flush=True)
                time.sleep(CHUNK_PAUSE)

    print("-" * 75, flush=True)
    print(f"Finished all {len(todo)} apps in {(time.time() - t0)/60:.1f} min.", flush=True)
else:
    print("Nothing to fetch — all apps already cached.", flush=True)
```

```python
# 4. Convert fetched results to DataFrame and merge into clean_data
scraped_df = pd.DataFrame([done[a] for a in apps if a in done])
cols_to_add = ["app", "matched_title", "app_id", "icon_url", "match_score", "title_score", "match_signals",
               "matched_query", "match_method", "runner_up_title", "match_status"]
scraped_df = scraped_df.reindex(columns=cols_to_add)

# Drop columns if re-running the cell so you don't get _x / _y duplicate columns
existing_cols = [c for c in cols_to_add if c != "app" and c in clean_data.columns]
if existing_cols:
    clean_data = clean_data.drop(columns=existing_cols)

clean_data = clean_data.merge(scraped_df, on="app", how="left")
```

### 4b. Fetch the full store data for each matched app (by `app_id`)

```python
# 4b. Load details cache, fetch the missing ones, merge on app_id
details_done = {}
if os.path.exists(DETAILS_CACHE_PATH):
    with open(DETAILS_CACHE_PATH, "r", encoding="utf-8") as f:
        for line in f:
            r = json.loads(line)
            if r.get("details_status") in ("ok", "not_found"):   # errors are retried on the next run
                details_done[r["app_id"]] = r

if FETCH_DETAILS:
    ids = (clean_data.loc[clean_data["match_status"].isin(DETAIL_STATUSES), "app_id"]
           .dropna().drop_duplicates().tolist())
    todo_ids = [i for i in ids if i not in details_done]
    print(f"App ids: {len(ids)} | Cached: {len(ids) - len(todo_ids)} | To fetch: {len(todo_ids)}")

    if todo_ids:
        t1, n_d = time.time(), 0
        n_chunks = -(-len(todo_ids) // CHUNK_SIZE)
        with open(DETAILS_CACHE_PATH, "a", encoding="utf-8") as cache:
            for ci, chunk in enumerate(chunked(todo_ids, CHUNK_SIZE), 1):
                with ThreadPoolExecutor(WORKERS) as pool:
                    futures = [pool.submit(fetch_details, i) for i in chunk]
                    for fut in as_completed(futures):
                        r = fut.result()
                        details_done[r["app_id"]] = r
                        cache.write(json.dumps(r, default=str) + "\n")
                        cache.flush()
                        time.sleep(0.2)
                n_d += len(chunk)
                el = time.time() - t1
                print(f"details chunk {ci}/{n_chunks} done | {n_d}/{len(todo_ids)} | ~{el / n_d * (len(todo_ids) - n_d) / 60:.1f} min left")
                if n_d < len(todo_ids):
                    time.sleep(CHUNK_PAUSE)

    details_df = pd.DataFrame([details_done[i] for i in ids if i in details_done])
    if not details_df.empty:
        # re-run safe: drop previously merged detail columns first
        stale = [c for c in details_df.columns if c != "app_id" and c in clean_data.columns]
        clean_data = clean_data.drop(columns=stale).merge(details_df, on="app_id", how="left")
```

```python
# 5. Save and inspect
clean_data.to_csv(OUTPUT_CSV, index=False)
print("\nMatch Status Breakdown:")
print(clean_data["match_status"].value_counts(dropna=False).to_string())
if "details_status" in clean_data.columns:
    print("\nDetails Status Breakdown:")
    print(clean_data["details_status"].value_counts(dropna=False).to_string())
print(f"\nSaved enriched dataset to: {OUTPUT_CSV} (Shape: {clean_data.shape})")
clean_data.head()
```
